```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
dpath = 'sample_data/Titanic-Dataset-fixed.csv'
df = pd.read_csv(dpath)
df.head()
```

```python
df = df[['Survived', 'Pclass', 'Age', 'Sex', 'Cabin']]
df.head()
```

```python
le = LabelEncoder() # create an object from LabelEncoder
df['Sex'] = le.fit_transform(df['Sex']) # encoding process
df['Cabin'] = le.fit_transform(df['Cabin']) # encoding process
df.head()
```

```python
std = StandardScaler()
df['Age'] = std.fit_transform(df[['Age']])
df.head()
```
